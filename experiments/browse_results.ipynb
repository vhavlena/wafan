# WAFan results browser
Lists the most recent result file per analysis, then lets you pick file(s) to inspect. All logic lives in `results_lib.py`.

```python
import results_lib as rl
from itables import init_notebook_mode
init_notebook_mode(all_interactive=True)

reports = rl.load_reports()
index_df = rl.report_index(reports)
latest = rl.latest_per_analysis(index_df)
latest
```

## Select report(s) to browse
Edit `selected` below (defaults to the latest report per analysis).

```python
selected = latest["path"].tolist()  # e.g. override: ["../results/wafan_files-20260709-095604.json"]
selected_reports = {p: reports[p] for p in selected}
list(selected_reports)
```

## Per-file summary

```python
files_df = rl.file_summary(selected_reports)
files_df
```

## Aggregate statistics

```python
rl.aggregate_stats(files_df)
```

## Unsupported features
Chain-record status: `ok` vs `unsupported_transform` / `unsupported_operator` / `unsupported_pattern`.

```python
rl.unsupported_details(selected_reports)
```

## Pair outcomes: ok / violates per (report, analysis), plus skipped / solver timeouts / solver errors

```python
rl.pair_outcome_counts(selected_reports)
```

## Rules violating the analyses
One row per pair record flagged as `violates` — the rule chains involved, the conf file, and (for contradiction) each chain's disposition.

```python
rl.violating_pairs(selected_reports)
```
